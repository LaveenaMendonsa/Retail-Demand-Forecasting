### 3.2 Merge Oil, Holiday, and Transaction Information

The following steps add date-level information. Holiday rows are aggregated by date before merging so that multiple holiday records do not multiply sales rows.

```python
# Merge daily oil prices into both datasets.
oil_daily = oil[["date", "dcoilwtico"]].drop_duplicates(subset="date")
df = df.merge(oil_daily, on="date", how="left", validate="many_to_one")
test = test.merge(oil_daily, on="date", how="left", validate="many_to_one")

# Aggregate holiday information to one row per date before merging.
holiday_daily = (holidays.groupby("date", as_index=False)
                 .agg(holiday_count=("description", "count"),
                      holiday_transferred_count=("transferred", "sum")))
holiday_daily["is_holiday"] = 1
df = df.merge(holiday_daily, on="date", how="left", validate="many_to_one")
test = test.merge(holiday_daily, on="date", how="left", validate="many_to_one")

# No transaction values are supplied for the test period; merge them into training only.
df = df.merge(transaction, on=["date", "store_nbr"], how="left", validate="many_to_one")

for frame in (df, test):
    frame["holiday_count"] = frame["holiday_count"].fillna(0)
    frame["holiday_transferred_count"] = frame["holiday_transferred_count"].fillna(0)
    frame["is_holiday"] = frame["is_holiday"].fillna(0).astype("int8")

print("Train after additional merges:", df.shape)
print("Test after additional merges:", test.shape)
display(df.head())
```

## 4. Feature Engineering

Create calendar features from the date column. These features help models learn weekly, monthly, and yearly sales patterns. The original `date` column is retained for sorting and validation, but it will not be passed directly into the model.

```python
def add_date_features(frame):
    frame = frame.copy()
    frame["year"] = frame["date"].dt.year
    frame["month"] = frame["date"].dt.month
    frame["day"] = frame["date"].dt.day
    frame["day_of_week"] = frame["date"].dt.dayofweek
    frame["week_of_year"] = frame["date"].dt.isocalendar().week.astype("int16")
    frame["is_weekend"] = (frame["day_of_week"] >= 5).astype("int8")
    return frame

df = add_date_features(df)
test = add_date_features(test)

print("Train shape:", df.shape)
print("Test shape:", test.shape)
display(df[["date", "year", "month", "day", "day_of_week", "week_of_year", "is_weekend"]].head())
```

## 6. Prepare Data for Machine Learning

We use a chronological validation split: the final 15 days of the training period are held out for validation. This is more appropriate for forecasting than randomly splitting dates.

The model-training cell uses a reproducible sample of up to 200,000 training rows to keep computation manageable on a personal computer. Validation is also sampled if it is very large. Therefore, results can differ slightly from experiments performed with different samples or settings.

```python
# Sort chronologically and define the final 15 days as validation.
df = df.sort_values("date").reset_index(drop=True)
validation_start = df["date"].max() - pd.Timedelta(days=14)

train_part = df[df["date"] < validation_start].copy()
valid_part = df[df["date"] >= validation_start].copy()

print("Validation starts:", validation_start)
print("Training rows before sampling:", len(train_part))
print("Validation rows before sampling:", len(valid_part))
print("Validation date range:", valid_part["date"].min(), "to", valid_part["date"].max())
```

```python
# Target and feature definitions.
TARGET = "sales"
# transactions are available for the training period but not for the test period,
# so exclude them from model features to keep train/test columns consistent.
DROP_COLUMNS = ["sales", "date", "id", "transactions"]
feature_columns = [col for col in train_part.columns if col not in DROP_COLUMNS]

# Categorical columns are encoded consistently using values learned from training only.
categorical_cols = train_part[feature_columns].select_dtypes(include=["object", "category"]).columns.tolist()
numeric_cols = [col for col in feature_columns if col not in categorical_cols]

print("Number of features:", len(feature_columns))
print("Categorical features:", categorical_cols)
print("Numeric feature count:", len(numeric_cols))
```

```python
# Take reproducible samples for faster local model training.
MAX_TRAIN_ROWS = 200_000
MAX_VALID_ROWS = 100_000

if len(train_part) > MAX_TRAIN_ROWS:
    train_sample = train_part.sample(n=MAX_TRAIN_ROWS, random_state=RANDOM_STATE)
else:
    train_sample = train_part.copy()

if len(valid_part) > MAX_VALID_ROWS:
    valid_sample = valid_part.sample(n=MAX_VALID_ROWS, random_state=RANDOM_STATE)
else:
    valid_sample = valid_part.copy()

X_train = train_sample[feature_columns].copy()
y_train = train_sample[TARGET].copy()
X_valid = valid_sample[feature_columns].copy()
y_valid = valid_sample[TARGET].copy()

print("Model training sample:", X_train.shape)
print("Validation sample:", X_valid.shape)
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
import joblib
import numpy as np

# Encode categorical columns; unknown categories in validation/test receive -1.
if categorical_cols:
    encoder = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)
    X_train[categorical_cols] = encoder.fit_transform(X_train[categorical_cols].astype(str))
    X_valid[categorical_cols] = encoder.transform(X_valid[categorical_cols].astype(str))
else:
    encoder = None

# Convert all features to numeric and replace any remaining missing values using training medians.
X_train = X_train.apply(pd.to_numeric, errors="coerce")
X_valid = X_valid.apply(pd.to_numeric, errors="coerce")
medians = X_train.median()
X_train = X_train.fillna(medians).astype("float32")
X_valid = X_valid.fillna(medians).astype("float32")

print("Missing values in X_train:", int(X_train.isna().sum().sum()))
print("Missing values in X_valid:", int(X_valid.isna().sum().sum()))
```

## 7. Train and Evaluate Regression Models

We compare Random Forest Regressor and HistGradientBoosting Regressor. MAE measures average absolute prediction error, RMSE penalizes larger errors more strongly, and R² measures explained variation. R² is **not** classification accuracy.

```python
models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=50,
        max_depth=15,
        min_samples_leaf=5,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),
    "HistGradientBoosting": HistGradientBoostingRegressor(
        max_iter=150,
        learning_rate=0.1,
        max_leaf_nodes=31,
        l2_regularization=1.0,
        random_state=RANDOM_STATE
    )
}

results = []
trained_models = {}

for model_name, model in models.items():
    print(f"Training {model_name} ...")
    model.fit(X_train, y_train)
    predictions = model.predict(X_valid)
    predictions = np.clip(predictions, 0, None)

    mae = mean_absolute_error(y_valid, predictions)
    rmse = np.sqrt(mean_squared_error(y_valid, predictions))
    r2 = r2_score(y_valid, predictions)

    results.append({
        "Model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })
    trained_models[model_name] = model

metrics_df = pd.DataFrame(results).sort_values("MAE")
display(metrics_df)
```

```python
# Visual comparison of validation metrics
metrics_long = metrics_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Error"
)

plt.figure(figsize=(9, 5))
sns.barplot(data=metrics_long, x="Model", y="Error", hue="Metric")
plt.title("Validation Error Comparison")
plt.tight_layout()
plt.show()
```

## 8. Prepare Test Features and Generate Predictions

The test dataset does not contain the target `sales`. We apply the same feature list, categorical encoder, and training-derived median values used during model preparation. The selected model below is HistGradientBoosting by default; change `SELECTED_MODEL` if you want to use Random Forest.

```python
# Prepare test features in the same column order as training.
X_test = test[feature_columns].copy()

if categorical_cols:
    X_test[categorical_cols] = encoder.transform(X_test[categorical_cols].astype(str))

X_test = X_test.apply(pd.to_numeric, errors="coerce")
X_test = X_test.fillna(medians).astype("float32")

assert list(X_test.columns) == list(X_train.columns)
print("Test feature matrix:", X_test.shape)
```

```python
SELECTED_MODEL = "HistGradientBoosting"

if SELECTED_MODEL not in trained_models:
    raise ValueError(f"Model {SELECTED_MODEL} was not trained in this notebook.")

final_model = trained_models[SELECTED_MODEL]
test_predictions = np.clip(final_model.predict(X_test), 0, None)

submission = pd.DataFrame({
    "id": test["id"],
    "sales": test_predictions
})

print("Submission shape:", submission.shape)
print("Missing values:", submission.isnull().sum().to_dict())
display(submission.head())
```

```python
# Save the prediction file in the project data folder.
submission_path = DATA_DIR / "submission_from_notebook.csv"
submission.to_csv(submission_path, index=False)
print("Saved submission to:", submission_path)
```

## 9. Save the Model Bundle

The bundle stores the fitted model, encoder, categorical column names, feature order, and median values needed for consistent preprocessing at inference time.

```python
models_dir = DATA_DIR.parent / "models"
models_dir.mkdir(parents=True, exist_ok=True)

model_bundle = {
    "model": final_model,
    "encoder": encoder,
    "categorical_cols": categorical_cols,
    "feature_columns": feature_columns,
    "medians": medians,
    "model_name": SELECTED_MODEL
}

model_path = models_dir / "retail_demand_model_from_notebook.pkl"
joblib.dump(model_bundle, model_path)
print("Saved model bundle to:", model_path)
```
